```python
import whisper
import torch

# تحميل نموذج Whisper Small على GPU
whisper_model = whisper.load_model("small", device="cuda")

print("✅ Whisper Small loaded successfully!")
print("Device:", next(whisper_model.parameters()).device)
```

```python
def transcribe_audio(audio_path):
    """
    Takes an audio file path and returns its Arabic transcription.
    """
    if not os.path.exists(audio_path):
        raise FileNotFoundError(f"Audio file not found: {audio_path}")

    result = whisper_model.transcribe(
        audio_path,
        language="ar",
        task="transcribe",
        fp16=True
    )

    transcript = result["text"].strip()
    return transcript


print("✅ transcribe_audio() is ready!")
```
